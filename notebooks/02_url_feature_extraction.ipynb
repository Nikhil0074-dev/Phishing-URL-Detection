```python
samples = [
    'https://github.com/explore',
    'http://secure-login.paypal-verify9.xyz/account/confirm.php',
    'http://192.168.1.20:8080/signin/update',
]
for url in samples:
    features = extract_features(url)
    print(url)
    print('  length=%d entropy=%.2f subdomains=%d suspicious_kw=%d https=%d'
          % (features['url_length'], features['url_entropy'],
             features['num_subdomains'], features['suspicious_keyword_count'],
             features['has_https']))
```
